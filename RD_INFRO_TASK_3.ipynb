```python
import zipfile
import os

zip_path = '/content/enron_spam_data.zip'

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall('/content/enron_spam_data')

print("Dataset extracted successfully!")
```

```python
import os

for root, dirs, files in os.walk('/content/enron_spam_data'):
    for file in files:
        if file.endswith('.csv'):
            print(os.path.join(root, file))
```

```python
import pandas as pd
df = pd.read_csv('/content/enron_spam_data/enron_spam_data.csv')
print("Dataset loaded successfully!")
print(df.head())
```

```python
print("Dataset Shape:")
print(df.shape)
print("/nColumn Names:")
print(df.columns)
print("/nDataset Information:")
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
df['Subject'] = df['Subject'].fillna('')

df['Message'] = df['Message'].fillna('')

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print(df['Spam/Ham'].value_counts())
```

```python
print(df['Spam/Ham'].unique())

print(df['Spam/Ham'].value_counts())
```

```python
df['Message_Length'] = df['Message'].apply(len)

print(df[['Message', 'Message_Length']].head())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df['Message_Length_Scaled'] = scaler.fit_transform(
    df[['Message_Length']]
)

print(df[['Message_Length', 'Message_Length_Scaled']].head())
```

```python
df.to_csv('cleaned_enron_spam_data.csv', index=False)

print("Task 3 Completed Successfully!")
print("Cleaned dataset saved successfully!")
```

```python
import zipfile
import os

with zipfile.ZipFile('cleaned_enron_spam_data.zip', 'w', zipfile.ZIP_DEFLATED) as z:
    z.write('cleaned_enron_spam_data.csv')

size = os.path.getsize('cleaned_enron_spam_data.zip') / (1024 * 1024)

print("Compressed file size:", round(size, 2), "MB")
```
